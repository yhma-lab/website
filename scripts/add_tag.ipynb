```python
import os
import yaml

# 指定 publication 目录
publication_dir = "/Users/eri/Desktop/bib/quantum_things"

new_tags = ["Quantum Gyroscope, Quantum Walks and Quantum Illumination"]

for folder in os.listdir(publication_dir):
    index_path = os.path.join(publication_dir, folder, "index.md")
    if os.path.exists(index_path):
        with open(index_path, "r", encoding="utf-8") as f:
            lines = f.readlines()

        # 找到 YAML block（--- 开头和结尾）
        if lines[0].strip() == "---":
            try:
                end_index = lines[1:].index("---\n") + 1
            except ValueError:
                print(f"跳过无效文件：{index_path}")
                continue

            yaml_block = "".join(lines[1:end_index])
            content_block = "".join(lines[end_index + 1 :])

            metadata = yaml.safe_load(yaml_block)

            # 合并原有 tags（如果有）
            existing_tags = metadata.get("tags", [])
            if isinstance(existing_tags, str):
                existing_tags = [existing_tags]

            metadata["tags"] = list(set(existing_tags + new_tags))

            # 重写文件
            with open(index_path, "w", encoding="utf-8") as f:
                f.write("---\n")
                yaml.dump(metadata, f, allow_unicode=True, sort_keys=False)
                f.write("---\n\n")
                f.write(content_block)

            print(f"✅ 已更新 tags: {index_path}")
```
